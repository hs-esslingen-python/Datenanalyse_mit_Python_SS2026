Wir wollen die species anhand der Länge und Breite der Blütenblätter (sepal, pedal) erkennen.

#### `pairplot`

`pairplot` ist in Seaborn eine Funktion, die für mehrere numerische Variablen **alle paarweisen Beziehungen auf einmal** zeigt.

Was du bekommst:
- Auf der Diagonale: Verteilung jeder einzelnen Variable (Histogramm/KDE).
- Außerhalb der Diagonale: Scatterplots für jede Variablen-Kombination.

Typischer Nutzen:
- Schneller Überblick über Zusammenhänge, Cluster und Ausreißer.
- Mit `hue='...'` kannst du Klassen farblich trennen (z. B. `species` bei Iris).

Beispiel:
```python
sns.pairplot(data=iris, hue='species')
```

Damit siehst du sofort, welche Merkmals-Kombinationen die Arten gut voneinander trennen.

````python
sns.pairplot(data=iris, hue='species');
````

## Datensatz 2:  miles per gallon dataset 

https://archive.ics.uci.edu/ml/datasets/auto+mpg


````python
mpg = sns.load_dataset('mpg')
mpg.head()
````

#### `relplot` 


`relplot` ist in Seaborn eine **Figure-level-Funktion** für die Darstellung von **Beziehungen zwischen Variablen** (relation plot), meist als:

- **Scatterplot** (`kind="scatter"`, Standard)
- **Linienplot** (`kind="line"`)

Typisch nutzt man es für Fragen wie:  
"Wie hängt `y` von `x` ab?" und kodiert weitere Merkmale über Farbe, Größe oder Stil.

In deinem Beispiel in  _06_Seaborn/Seaborn - Beispiele.ipynb  mit

- `x='horsepower'`
- `y='mpg'`
- `hue='origin'`
- `size='weight'`

erzeugt `sns.relplot(...)` einen **Bubble-Scatterplot**:  
x-Achse = PS, y-Achse = Verbrauch (mpg), Farbe = Herkunft, Punktgröße = Gewicht.

Kurz: `relplot` ist der flexible Standard in Seaborn, um **Zusammenhänge zwischen numerischen Variablen** sichtbar zu machen.





````python
sns.relplot(data=mpg, x='horsepower', y='mpg', hue='origin', size='weight', 
            sizes=(40, 400), alpha=0.5, palette="muted", height=5);
````
