```python
import numpy as np
import pandas as pd
```

### Erzeugen eines _Pandas DataFrame-Objektes_ 

**Beispiel**: Tabelle mit 4 Zeilen und 5 Spalten und den Zahlen von 1 bis 20

```python
my_index   = ['Zeile1','Zeile2','Zeile3','Zeile4']
my_columns = ['Spalte1','Spalte2','Spalte3','Spalte4','Spalte5']
my_data    = np.arange(1,len(my_index)*len(my_columns)+1).reshape(len(my_index),len(my_columns))
df = pd.DataFrame(data=my_data,index=my_index,columns=my_columns)
df
```

```python
my_data
```

#### Zugriff auf eine Spalte, Rückgabe erfolgt als eine Pandas Series 
Beispiel "Spalte1"

```python
df["Spalte1"]
```

alternative mit Punkt-Notation, wenn die Bezeichnung der Spalte dies zulässt:
<br> 
Spaltenname darf keine Leerzeichen enthalten und darf sich nicht mit bereits belegten Attributen überschneiden, z.B. index, column.

```python
df.Spalte1
```

#### Zugriff auf eine Spalte, Rückgabe erfolgt als ein Pandas DataFrame 
Beispiel "Spalte1"
<br>
Es wird eine Liste mit einem Element übergeben.

```python
df[["Spalte1"]]
```

#### Zugriff auf mehrere ausgewählte Spalten; Rückgabe erfolgt als ein Pandas DataFrame 
Beispiel "Spalte1" und "Spalte3"
<br>
Die Namen der Spalten werden als eine Liste übergeben

```python
df[ ["Spalte1", "Spalte3"] ]
```

#### Übergabe einer Maske (Array mit True/False Werte z.B. aus einer Bedingung)
Beispiel: Erzeugen einer Maske für die Zeilen, in der die Werte der Spalte "Spalte1" größer 5 sind

```python
mask = df.Spalte1 > 5
mask
```

```python
df[mask]
```

### Hintereinanderschalten von Maske und Spaltenauswahl
Beispiel: wie oben, aber nur die "Spalten1" ausgeben; Rückgabe als eine Pandas Series

```python
mask = df.Spalte1 > 5
#df[mask]["Spalte1"]
df.loc[mask, "Spalte1"]
```

Beispiel: wie oben, aber jetzt die "Spalten1" und die "Spalte3" auswählen.
<br> 
Mehrere Spalten werden als eine Liste übergeben; 
<br>
Rückgabe erfolgt als ein Pandas DataFrame.

```python
mask = df.Spalte1 > 5
#df[mask][["Spalte1","Spalte3"]]
df.loc[mask, ["Spalte1","Spalte3"]]
```

####  Zugriff auf mehrere Zeilen mittels Slicing und Integer Index; Rückgabe Pandas DataFrame 
Beispiel: Auswahl der Zeilen 2 bis 4.
Index 0 entspricht Zeile 1; Index 1 entspricht Zeile 2, Index 4 entspricht Zeile 5.
<br>
Beim Integerindex ist Start enthalten, aber Stop ist nicht mehr enthalten.

```python
df[1:4]
```

### `.loc[ ]`   (Zugriff mit Labeln)

#### Zugriff auf eine Zeile; Rückgabe Pandas Series
Beispiel: "Zeile1"

```python
df.loc["Zeile1"]
```

#### Zugriff auf eine Zeile; Rückgabe als Pandas DataFrame
Beispiel: "Zeile1" 
<br>
Übergabe einer Liste!

```python
df.loc[["Zeile1"]]
```

#### Zugriff auf mehrer ausgewählte Zeilen mittels Liste; Rückgabe als Pandas DataFrame
Beispiel: "Zeile1", "Zeile3" und "Zeile4"

```python
df.loc[ ["Zeile1", "Zeile3", "Zeile4"] ] 
```

#### Zugriff auf Zeilen mit Slicing; Rückgabe als Pandas DataFrame
Beim Slicing mit Zeilennamen (Label) sind Start und Stop enthalten.
<br>
Beipiel: "Zeile1" bis "Zeile3" (inklusive)

```python
df.loc[ "Zeile1" : "Zeile3" ] 
```

Beispiel: Ausgeben aller Zeilen; 
<br> 
Start und Stop können weggelassen werden. Es muss nur der Doppelpunkt angegeben werden.

```python
df.loc[:] 
```

Beispiel Schrittweite (Step), 
<br>
Ausgabe der Zeilen mit gerader Zeilennummer bis zum Ende der Zeilen. 
<br>
Das Ende (Stop) kann weggelassen werden.

```python
df.loc['Zeile2'::2]
```

#### Zugriff auf eine Spalten; Rückgabe als Pandas Series
Beispiel: Ausgabe aller Zeilen der "Spalte1"

```python
df.loc[:,"Spalte2"]
```

#### Zugriff auf eine Spalten; Rückgabe als Pandas DataFrame

Beispiel: Ausgabe aller Zeilen der "Spalte1"
<br>
Die Spalte wird als Liste übergeben.

```python
df.loc[:,["Spalte2"]]
```

#### Zugriff auf mehrere ausgewählte Spalten; Rückgabe als Pandas DataFrame 
Beispiel: Rückgabe der "Spalte1", "Spalte3" und "Spalte5" 
<br>
Die Spalten werden in einer Liste übergeben.

```python
df.loc[ : , ["Spalte1", "Spalte2", "Spalte3"] ] 
```

#### Zugriff unter Verwendung einer Maske
Beispiel: Ausgabe aller Zeilen in der die "Spalte1" einen Wert größer 5 hat

```python
mask = df["Spalte1"] > 5
mask
```

```python
df.loc[mask]
```

#### Zugriff unter Verwendung einer Maske - zusätzlich Ausgabe auf ausgewählte Spalten beschränken
Beispiel: wie oben nur dass nur "Spalte1" und "Spalte4" ausgegeben werden. 

```python
mask = df["Spalte1"] > 5
df.loc[mask,["Spalte1", "Spalte4"]] 
```

#### Zugriff auf ausgewählte Zeilen und Spalten; Rückgabe als Pandas DataFrame
Übergabe der Namen in Form einer Liste oder als Slicing (Indexer).
Start und Stop können weggelassen, wenn sie sich auf Anfang oder Ende der Liste beziehen.
Angabe einer Schrittweite ist möglich.

Beispiel: Zeilen und Spalten durch Listen ausgewählt.

```python
df.loc[ ["Zeile1", "Zeile3"] , [ "Spalte2", "Spalte3"] ]
```

Beispiel: Zeilen durch Slicing und Spalten durch Liste ausgewählt.

```python
df.loc[ "Zeile1": "Zeile3" , [ "Spalte2", "Spalte3"] ]
```

Beispiel: Zeilen und Spaltzen durch Slicing ausgewählt.

```python
df.loc[ "Zeile1": "Zeile3" , "Spalte2"::2 ]
```

### `.iloc[ ]`    (Zugriff mit Integer-Index -  integer-location based indexing)
**Bitte beachten:** Beim Slicing mit Integer-Index ist der Start enthalten, aber der Stop ist nicht enthalten.

#### Zugriff auf ausgewählte Zeilen über eine Liste; Rückgabe als Pandas DataFrame
Beispiel: Rückgabe von "Zeile1" (Index 0) und "Zeile3" (Index 2).

```python
df.iloc[ [0, 2] ] 
```

#### Zugriff auf ausgewählte Zeilen über Slicing; Rückgabe als Pandas DataFrame
**Bitte beachten:** Beim Slicing mit Integer-Index ist der Start enthalten, aber der Stop ist nicht enthalten.
<br>
Beispiel: Rückgabe von "Zeile1" (Index 0) bis "Zeile4" (Index 3).
<br> 
Für das Ende (Stop) muss der nächste Index 4 angebene werden !!!
 

```python
df.iloc[ 0 : 4] 
```

#### Zugriff auf ausgewählte Spalten über eine Liste; Rückgabe als Pandas DataFrame
Beispiel: Rückgabe der "Spalte1" (Index 0), "Spalte3" (Index 2) und "Spalte4" (Index 3).

```python
df.iloc[ : , [0, 2, 3] ]
```

#### Zugriff auf ausgewählte Zeilen und Spalten über Listen und Slicing -> Pandas DataFrame
Beispiel: Rückgabe der "Zeile1" (Index 0) und "Zeile3" (Index 2) nur mit der "Spalte2" (Index 1) und "Spalte4" (Index 3).

```python
df.iloc[ [0, 2] , [1, 3] ]
```

Beispiel: Rückgabe der "Zeile1" (Index 0) bis "Zeile4" (Index 3) nur mit der "Spalte2" (Index 1) und "Spalte4" (Index 3).

```python
df.iloc[0:4 , [1, 3] ]
```

Beispiel: Rückgabe der "Zeile1" (Index 0) bis "Zeile4" (Index 3) nur mit der "Spalte2" (Index 1) bis "Spalte4" (Index 3).

```python
df.iloc[ 0:4 , 1:4 ]
```
