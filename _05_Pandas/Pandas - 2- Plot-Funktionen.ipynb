## Ersten eines DataFrames ("Großer Datensatz") 

Im folgenden wird ein Datensatz mit 4 Spalten A, B, C und D und 100 Zeilen erstellt der mit Zufallzahlen gefüllt wird.

```python
N = 100
np.random.seed(7) # Zufallsgenerator liefert reproduzierbare Werte
data = np.random.randn(N,4)  # Zufallszahlen
df = pd.DataFrame(data = data, columns=['A','B','C','D'])
df.head()
```

### Deskriptive Statistik der erzeugten Daten

Die zurückgegebene Zusammenfassung enthält Informationen wie Anzahl, Durchschnitt, Standardabweichung, Minimum, 25. Perzentil, Median, 75. Perzentil und Maximum für jede numerische Spalte im DataFrame. 
<br>
Diese Informationen sind nützlich, um einen schnellen Überblick über die Verteilung der Daten zu erhalten und potenzielle Ausreißer oder fehlende Werte zu identifizieren.

```python
df.describe()
```

#### Was ist ein Perzentil?

Ein **Perzentil** ist ein statistisches Maß, das verwendet wird, um die Position eines bestimmten Wertes in einer geordneten Datenreihe zu bestimmen. Es gibt verschiedene Perzentile, die verwendet werden können, wie z.B. das 25. Perzentil (auch bekannt als das **untere Quartil**), das 50. Perzentil (auch bekannt als der **Median**) und das 75. Perzentil (auch bekannt als das **obere Quartil**).

Das Perzentil gibt an, welcher Anteil der Datenwerte unterhalb eines bestimmten Wertes liegt. Zum Beispiel liegt das 25. Perzentil (unteres Quartil) an der Stelle, an der 25% der Datenwerte kleiner oder gleich diesem Wert sind. Das 50. Perzentil (Median) liegt an der Stelle, an der 50% der Datenwerte kleiner oder gleich diesem Wert sind. Das 75. Perzentil (oberes Quartil) liegt an der Stelle, an der 75% der Datenwerte kleiner oder gleich diesem Wert sind.

Perzentile sind nützlich, um die **Verteilung der Daten** zu verstehen und **Ausreißer oder extreme Werte zu identifizieren**. Sie ermöglichen es, die Daten in verschiedene Bereiche zu unterteilen und die Position eines bestimmten Wertes relativ zur Gesamtdatenmenge zu bestimmen.

### Linienplot `.plot()`

### Histogramm einer Spalte

Die [`.plot.hist()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.plot.hist.html)-Methode stellt ein Histogram der ausgewählten Spalten dar.

Was stellt ein Histogram dar?

[Wikipedia - Histogramm](https://de.wikipedia.org/wiki/Histogramm)

Ein Histogramm zeigt, wie die Daten über einen bestimmten Wertebereich verteilt sind und gibt Aufschluss über die Häufigkeit oder den Anteil der Daten in verschiedenen Bereichen.

Bei der Erstellung eines Histogramms wird der Wertebereich in eine Reihe von **Bins oder Intervallen** unterteilt. 
Jeder Bin repräsentiert einen bestimmten Bereich von Werten, und die Höhe des Balkens im Histogramm gibt an, wie viele Datenpunkte in diesem Bin fallen.

Ein Histogramm kann verwendet werden, um verschiedene Aspekte der Verteilung zu visualisieren, wie zum Beispiel die Form der Verteilung (z.B. symmetrisch, schief, bimodal), das Vorhandensein von Ausreißern oder ungewöhnlichen Werten, die Konzentration der Daten in bestimmten Bereichen und die allgemeine Streuung der Daten.

Es ist wichtig zu beachten, dass die Interpretation eines Histogramms von der Anzahl der Bins abhängt, die für die Darstellung verwendet werden. Eine zu grobe Unterteilung der Daten kann wichtige Details der Verteilung verbergen, während eine zu feine Unterteilung zu einer Überbetonung von Rauschen oder kleinen Schwankungen führen kann. Die Wahl der richtigen **Anzahl von Bins** ist daher eine wichtige Überlegung bei der Erstellung eines Histogramms für kontinuierliche Daten.


```python
df['A'].plot.hist()
plt.show()                           # Diagramm anzeigen
```

Die Anzahl der Bins ist standardmäßig 10 eingestellt.
Diese kann durch den Parameter `bins=` z.B. auf 20 verändert werden.

```python
df['A'].plot.hist(bins=20)
plt.show()                           # Diagramm anzeigen
```

```python
df['A'].plot.hist(bins=100)
plt.show()                           # Diagramm anzeigen
```

Alternativ kann ein Histogram über den `kind='hist'`-Parameter ausgewählt werden.

```python
df['A'].plot(kind='hist',bins=20)
plt.show()                           # Diagramm anzeigen
```

### Histogramme mehrerer Spalten

Darstellung mehrere Histogramme übereinander in einem Diagramm, zum Vergleich der Verteilungen:

```python
df[['A','B']].plot(kind='hist',bins=20,alpha=0.5)
plt.show()                           # Diagramm anzeigen
```

### Boxplot einer Spalte

Die [`.plot.box()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.plot.box.html)-Methode stellt einen Boxplot

Was stellt ein Boxplot dar?

[Wikipedia - Box-Plot](https://de.wikipedia.org/wiki/Box-Plot)

Siehe auch Folie zum Thema "Boxplot (Box-Whisker-Plot oder Kastengrafik)"

Ein Boxplot stellt Median, die zwei Quartile und die beiden Extremwerte dar.

Ein Boxplot zeigt die Verteilung der Daten in Bezug auf verschiedene **statistische Kennzahlen** wie den Median, das erste und dritte Quartil, "Whiskers" und potenzielle Ausreißer.

Ein Boxplot besteht aus einer Box, die den **Interquartilbereich (IQR)** darstellt, der den Bereich zwischen dem ersten Quartil (25. Perzentil) und dem dritten Quartil (75. Perzentil) umfasst. Innerhalb der Box befindet sich eine Linie, die den **Median** repräsentiert. Die sogenannten **"Whiskers"** (Striche) erstrecken sich von der Box aus und repräsentieren den Bereich der Daten, der nicht als Ausreißer betrachtet wird. Ausreißer werden als einzelne Punkte außerhalb der Whiskers dargestellt.

Ein Boxplot ermöglicht es, schnell wichtige Informationen über die Verteilung der Daten zu erfassen, wie z.B. die Lage des Medians, die Streuung der Daten und das Vorhandensein von Ausreißern. Es ist besonders nützlich, um Vergleiche zwischen verschiedenen Gruppen oder Kategorien von Daten zu ziehen.


```python
df.loc[10,'A'] =3 # Ausreißer hinzufügen
df.loc[11,'A'] =-2 # Ausreißer hinzufügen
df.loc[12,'A'] =-2.2 # Ausreißer hinzufügen
```

```python
df['A'].describe()
```

```python
df['A'].plot.box()
plt.grid()
plt.show()                           # Diagramm anzeigen
```

Alternativ kann ein Boxplot über den `kind='box'`-Parameter ausgewählt werden.

```python
df['A'].plot(kind='box')
plt.show()                           # Diagramm anzeigen
```

### Boxplote mehrerer Spalten

Mehrere Boxplote nebeneinander in einem Diagramm, zum Vergleich der Verteilungen:

```python
df[['A','B','C','D']].plot(kind='box')
plt.show()                           # Diagramm anzeigen
```

### Plotten eines ganzen Datensatzes als Linienplots

```python
df.plot()
plt.show()                           # Diagramm anzeigen
```

Standardmäßig werden die Linien der Spalten in ein Diagramm geplottet.

Durch den Parameter `subplots=True` werden die einzelnen Spalten des Datensatzes in separate Subplots geplottet.

Durch Parameter, die wir von Matplotlib kennen, kann die Darstellung der Plots verändert werden.
- Größe der Figure: `figsize=(16,8)`
- Markerstyle: `marker='.'`
- Strichstärke der Linien: `linewidth=1.5` 
- Liniendarstellung : `linestyle=':'`
- Durchsichtigkeit der Farbflächen: `alpha=0.5`

Die `.plot()`-Methode gibt erzeugten axes-Objekte für die weitere Bearbeitung zurück. 

```python
axs = df.plot(subplots=True,figsize=(16,8),marker='.', linestyle=':', linewidth=1.5, alpha=0.5)

# in jedes Axes-Objekt zeichen mit ein Gitter ein
for ax in axs:
    ax.grid(True)
plt.show()                           # Diagramm anzeigen
```

## Scatterplot (Streudiagramm)

Mit der [`.plot.scatter()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.plot.scatter.html)-Methode kann ein Scatterplot erstellt werden.

Was stellt ein Scatteplot dar?

[Wikipedia -Streudiagramm](https://de.wikipedia.org/wiki/Streudiagramm)

Parameter:
- `x=` x - Achse
- `y=` y - Achse
- `s=` Marker-Größe
- `c=` Marker-Farbe

Darstellung von zwei Größen: Spalte 'B' über Spalte 'A' plotten

```python
df.plot.scatter(x='A',y='B')
plt.show()                           # Diagramm anzeigen
```

```python
df.plot.scatter(x='B',y='C')
plt.show()                           # Diagramm anzeigen
```

Darstellung von drei Größen: Spalte 'B' über Spalte 'A' plotten; Spalte 'C' über die Farbe des Punktes darstellen:

```python
df.plot.scatter(x='A',y='B',c=df['C'],cmap='coolwarm',figsize=(10,8))
plt.show()                           # Diagramm anzeigen
```
