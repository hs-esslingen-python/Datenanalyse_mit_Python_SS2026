### _Auto MPG_ Datensatz laden

Beschreibung der Spalten des  _Auto MPG_ Datensatzes:


|Spalte|Bezeichnung|Beschreibung|Einheit|Datentyp|
|---|---|---|---|---|
|1. |mpg         | Benzinverbrauch des Auto | miles per gallon | continuous|
|2. |cylinders   | Zylinderanzahl des Motors| -                |  multi-valued discrete|
|3. |displacement| Hubraum des Motors       | kubik inch       | continuous|
|4. |horsepower  | Leistung des Motors       | Pferdestärken PS | continuous|
|5. |weight      | Gewicht des Autos        | Pfund            | continuous|
|6. |acceleration| Beschleunigung des Autos | Zeit in Sekunden für Stand bis 60 Meilen pro Stunde| continuous|
|7. |model year  | Baujahr                  | -                | multi-valued discrete |
|8. |origin      | Herkunft der Automarke   | -                | multi-valued discrete
|9. |name        | Automarke und Modellname | -                | string (unique for each instance)|

Referenz:
Details zur Beschreibung der Spalten aus [Sasche Serafimovski; Auto-MPG. Einfluss verschiedenster Kriterien auf den Benzinverbrauch eines Autos; 2018](https://www.grin.com/document/442592)


````python
df = sns.load_dataset('mpg')
df.head()
````

````python
df['my_index'] = [f"car_{k}" for k in range(len(df))]
df=df.set_index('my_index')
df.head()
````
