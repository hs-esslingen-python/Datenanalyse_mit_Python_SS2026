# Pandas - Zusammenführen von DataFrames

Methoden zum "Zusammenführen" von mehreren _Pandas DataFrames_:

- `pd.concat()`, `.append()`  - "Aufeinander/Nebeneinander-Stapeln" von DataFrames
- `pd.merge()`, `.join()`   - Zusammenfügen von DataFrame mittels Abgleich anhand Spalten oder Indices


Die Top-Level-Funktion `pd.concat()` und die `.append()`-Methode weisen eine ähnliche Funktionalität auf, dies gilt auch für die Top-Level-Funktion `pd.merge()` und die `.join()`-Methode.

Die Top-Level-Function `pd.concat()` und `pd.merge()` bieten aber eine größere Flexibilität.

Die `.append()`-Methode ist bereits abgekündigt.

```python
import numpy as np 
import pandas as pd
```

## "Aufeinander/Nebeneinander-Stapeln" von DataFrames 

Wir erzeugen uns drei DataFrames `df1`, `df2` und `df3`, die wir zusammenbauen wollen.

Alle drei DataFrames haben die Spalten 'A' bis 'D':

```python
data1 = {'A':['A1','A2','A3','A4'],
         'B':['B1','B2','B3','B4'],
         'C':['C1','C2','C3','C4'],
         'D':['D1','D2','D3','D4'],
        }
index1 = ['I1','I2','I3','I4']
df1 = pd.DataFrame(data=data1,index=index1)
df1
```

```python
data2 = {'A':['A5','A6','A7','A8'],
         'B':['B5','B6','B7','B8'],
         'C':['C5','C6','C7','C8'],
         'D':['D5','D6','D7','D8'],
        }
index2 = ['I5','I6','I7','I8']
df2 = pd.DataFrame(data=data2,index=index2)
df2
```

```python
data3 = {'A':['A9','A10','A11','A12'],
         'B':['B9','B10','B11','B12'],
         'C':['C9','C10','C11','C12'],
         'D':['D9','D10','D11','D12'],
        }
index3 = ['I9','I10','I11','I12']
df3 = pd.DataFrame(data=data3,index=index3)
df3
```

### Top-Level Funktion `pd.concat()`

Der Top-Level Funktion [`pd.concat()`](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.concat.html) wird eine Liste mit DataFrames übergeben, die "gestapelt" werden sollen.

Zuerst aufeinander stapeln (parameter `axis=0` - default)

```python
pd.concat([df1,df2,df3])
```

Jetzt nebeneinander stapeln (parameter `axis=1`)

```python
pd.concat([df1,df2,df3],axis=1)
```

die nicht ausgefüllten Plätze werden mit 'NaN' (Not a Number) gefüllt

### Was passiert wenn die DataFrame unterschiedliche Spalten haben?

Wir erzeugen einen vierten Dataframe `df4` mit den Spalten 'A', 'B','E' und 'F':

```python
data4 = {'A':['A9','A10','A11','A12'],
         'B':['B9','B10','B11','B12'],
         'E':['C9','C10','C11','C12'],
         'F':['D9','D10','D11','D12'],
        }
index4 = ['I9','I10','I11','I12']
df4 = pd.DataFrame(data=data4,index=index4)
df4
```

```python
pd.concat([df1,df2,df4])
```

Das Zusammenfügen erfolgt nach den Spalten; fehlende Daten werden durch 'NaN' aufgefüllt.

## Zusammenfügen von DataFrame mittels Abgleich anhand Spalten oder Indices

Wir erzeugen uns zwei DataFrames, einen mit Produkten im Supermarkt und einen anderen mit einem Einkaufskorb.

Die Spalte 'Produkt_Id' verbindet beide DataFrames.

```python
Produkte_Supermarkt = pd.DataFrame( {'Produkt_Id'    : ['id1','id2','id3','id4'],
                                     'Produkt_Name'  : ['Milch','Käse','Wasser','Äpfel'],
                                     'Produkt_Preis' : [   1.39,  1.99,    0.29,   3.64],})
Produkte_Supermarkt
```

```python
Einkaufskorb = pd.DataFrame({'Lfd-Nr'    : [      1,      2,      3,      4,        5],
                             'Produkt_Id': [  'id2',  'id1',  'id4', 'id10',    'id4'],
                             'Anzahl'    : [      3,      2,      1,      5,        2],
                             'Uhrzeit'   : [ '8:00', '8:00', '8:00', '9:00', '10:00' ]})
Einkaufskorb
```

### Top Level-Funktion `pd.merge()`

Mi der Top-Level Funktion [`pd.merge()`](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.merge.html) wird ein neuer DataFrame aus zwei DataFrames anhand des Abgleich von Schlüsseln
zusammengebaut.

"database-style join style"

```python
pd.merge(Einkaufskorb,Produkte_Supermarkt)
```

Die Spalte 'Produkt_Id' ist sowohl im DataFrame 'Einkaufskorb' als im DataFrame 'Produkte_Supermarkt' enthalten.
Die Top Level-Funktion `pd.merge()` baut in der Default-Einstellung einen neuen DataFrame mit Zeilen, wo die 'Produkt_Id' in den Zeilen der beiden DataFrames jeweils übereinstimmen. Zeilen, die nur in einem der DataFrame vorkommen, werden nicht übernommen
Diese Zusammenfassen wird auch als **"inner joint"** bezeichnet.

Es gibt noch drei weiter Möglichkeiten des Zusammenfügens:

- **outer joint**
- **left joint**
- **right joint**



```python
pd.merge(Einkaufskorb,Produkte_Supermarkt, how='outer')
```

```python
pd.merge(Einkaufskorb,Produkte_Supermarkt, how='left')
```

```python
pd.merge(Einkaufskorb,Produkte_Supermarkt, how='right')
```

### `.join()`-Methode 

Mit der [`.join()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.join.html)-Methode
werden zwei DataFrame
anhand von:
- gleichnamigen Spalten oder
- dem Index
zusammengefasst.

Im DataFrame 'Produkte_Supermarkt' wird die Spalte 'Produkt_Id' als Index gesetzt:

```python
Produkte_Supermarkt_mit_Produkt_Id = Produkte_Supermarkt.set_index('Produkt_Id')
Produkte_Supermarkt_mit_Produkt_Id
```

Jetzt wird der DataFrame 'Einkaufskorb' anhand der Spalte 'Produkt_Id' mit DataFrame 'Produkte_Supermarkt_mit_Produkt_Id' und dessen Index zusammengeführt.

```python
Einkaufskorb.join(Produkte_Supermarkt_mit_Produkt_Id,on='Produkt_Id')
```
