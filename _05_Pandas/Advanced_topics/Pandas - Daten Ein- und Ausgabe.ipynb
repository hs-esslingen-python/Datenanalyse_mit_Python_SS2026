#### DatenFrame `df` mit Integerzahlen erstellen
Wir erstellen einen _Pandas DataFrame()_ mit 3 Spalten und 4 Zeilen und den Integeruahlen von 0 bis 11:

```python
columns = ['Spalte1', 'Spalte2', 'Spalte3']
index   = ['Zeile1', 'Zeile2', 'Zeile3', 'Zeile4'] 
data = np.arange(len(index)*len(columns)).reshape(len(index),len(columns))
df = pd.DataFrame(data=data, columns=columns, index= index)
df
```

Welchen Name hat der Index?

```python
print(df.index.name)
```

Aktuell ist kein Name vergeben.

Wir können dem Index einen Namen z.B. `my_index` geben:

```python
df.index.name = 'my_index'
```

```python
print(df.index.name)
```

Der Name des Index taucht oberhalb der Bezeichnung der Zeilen auf:

```python
df
```

#### DataFrame `df_float` mit Fließkommazahlen

Wir erstellen einen weiteren _Pandas DataFrame()_ mit 3 Spalten und 5 Zeilen und zufälligen Fließkommazahlen:

```python
df_float = pd.DataFrame(data=np.random.randn(5,3),columns=['M1','M2','M3'],index=['N1','N2','N3','N4','N5'])
df_float.index.name = 'my_index_float'
df_float
```

### CSV Datei

Mit der [`.to_csv`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.to_csv.html)-Methode
wird der DataFrame in eine CSV - Datei geschrieben (Export).

Der Index-Spalte wurde durch den Parameter `index_label='my_index'` ein Namen geben: 

`df.to_csv(CSV_Dateiname,index_label='my_index')`

```python
CSV_Dateiname = r'df.csv'
df.to_csv(CSV_Dateiname)
```

Mit der Funktion [`pd.read_csv()`](https://pandas.pydata.org/docs/reference/api/pandas.read_csv.html)
kann eine csv-Datei in einen DatenFrame eingelesen werden: 


```python
df2 = pd.read_csv(CSV_Dateiname) 
df2
```

Der Index `my_index` wurde als Spalte eingelesen.

Um die Spalte `my_index` als Index einzulesen, muss der Parameter `index_col=` auf den Name der Spalte (hier: 'my_index') gesetzt werden. 

```python
df2 = pd.read_csv(CSV_Dateiname, index_col='my_index')
df2
```

---
### _'Tabular separated data'_ - Datei 

Mit dem Parameter `sep='\t'` kann das Trennzeichen (Separator) der Einträge auf einen Tabulator umgestellt werden.

_'Tabular separated data'_ - Datei schreiben: `.to_csv(...,sep='\t')`


```python
Tab_Dateiname = r'df.tab'
df.to_csv(Tab_Dateiname,sep='\t')
```

_Tabular separated data_ - Datei wieder einlesen: `pd.read_csv(...,sep='\t')`

```python
df3 = pd.read_csv(Tab_Dateiname,sep='\t',index_col='my_index')
df3
```

---
### CSV - Datei mit Semicolon als Trennzeichen

CSV - Datei mit Semicolon schreiben `.to_csv(..,sep=';')`

```python
CSV_Semocolon_Dateiname = r'df_semicolon.csv'
df.to_csv(CSV_Semocolon_Dateiname,sep=';')
```

CSV - Datei mit Semicolon als Trennzeichen wieder einlesen 
`pd.read_csv(...,sep=';')`

```python
df4 = pd.read_csv(CSV_Semocolon_Dateiname,sep=';',index_col='my_index')
df4
```

#### Fließkommazahlen mit einem Komma als Dezimaltrennzeichen

Beim Schreiben der csv-Datei werden mit dem Paramter `decimal=","` die Fließkommazahlen mit einem Komma als Dezimaltrennzeichen anstatt mit einem Punkt geschrieben.

Bei Einlesen der csv-Datei werden mit dem Paramter `decimal=","` Kommas als Dezimaltrennzeichen interpretiert.
Ohne den Parameter werden die Floatzahlen als Zeichenkette eingelesen.

<!--
https://de.wikipedia.org/wiki/Dezimaltrennzeichen
https://living-sun.com/python/730337-convert-commas-to-dots-within-a-dataframe-python-pandas.html
-->

CSV mit Semicolon als Trennzeichen und Komma als Dezimaltrennzeichen schreiben:

```python
CSV_Semicolon_float_Dateiname = r'df_semicolon_float.csv'
df_float.to_csv(CSV_Semicolon_float_Dateiname,sep=';',decimal=",")
```

CSV mit Semicolon als Trennzeichen und Komma als Dezimaltrennzeichen wieder leseb:

```python
df2_float = pd.read_csv(CSV_Semicolon_float_Dateiname,sep=';',index_col='my_index_float',decimal=",")
print(df2_float.dtypes)
df2_float
```

Wird beim Einlesen der Parameter `decimal=","` nicht angegeben, werden die Zahlen als Zeichenketten ('object') eingelesen und nicht als 'float64'.

```python
df2_float = pd.read_csv(CSV_Semicolon_float_Dateiname,sep=';',index_col='my_index_float')
print(df2_float.dtypes)
df2_float
```

Durch das Laden der Python Pakete kann überprüft werden, ob diese korrekt installiert wurden:

Mit der [`to_excel`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.to_excel.html)-Methode wird ein DataFrame als Excel-Tabelle abgespeichert:

#### im XLS-Format (ist aber abgekündigt)

```python
Excel_xls_Dateiname = r'df.xls'
df.to_excel(Excel_xls_Dateiname,engine='openpyxl')
```

```python
df5 = pd.read_excel(Excel_xls_Dateiname, index_col='my_index')
df5
```

#### im XLSX Format

Optionale Angabe eines Names für das Tabellenblatts mit dem Parameter `sheet_name=`

```python
Excel_xlsx_Dateiname = r'df.xlsx'
df.to_excel(Excel_xlsx_Dateiname, sheet_name='My_DataFrame')
```

```python
df6 = pd.read_excel(Excel_xlsx_Dateiname, index_col='my_index')
df6
```

Mit [`pd.ExcelWriter()`](https://pandas.pydata.org/docs/reference/api/pandas.ExcelWriter.html#pandas.ExcelWriter)
läßt sich eine Excel-Mappe mit mehreren Tabellenblättern anlegen.

Mit einem _Context-Manager_ `with pd.ExcelWriter(Excel_xlsx_mehrere_Tabellen_Dateiname) as writer:` wird die Excel-Mappe angelegt und dann im einrückten Abschnitt beschrieben. 
Zuerst mit dem Integer-DataFrame und in der darauf folgenden Zeile mit dem Float-DataFrame.
Die Excel-Mappe wird beim Verlassen des _Context-Managers_ geschlossen.

```python
Excel_xlsx_mehrere_Tabellen_Dateiname = r'df_mehrere_Tabellen_.xlsx'
with pd.ExcelWriter(Excel_xlsx_mehrere_Tabellen_Dateiname) as writer:  
    df.to_excel(writer, sheet_name='Sheet_name_1_Integer')
    df_float.to_excel(writer, sheet_name='Sheet_name_2_Float')
```

Lesen des erstes Tabellenblattes:

```python
df7 = pd.read_excel(Excel_xlsx_mehrere_Tabellen_Dateiname,index_col='my_index')
df7
```

Lesen des zweiten Tabellenblattes durch explizite Angabe des Names des Tabellenblattes durch den `Parametersheet_name='Sheet_name_2_Float'`

```python
df8 = pd.read_excel(Excel_xlsx_mehrere_Tabellen_Dateiname,sheet_name='Sheet_name_2_Float',index_col='my_index_float')
df8
```

Um ein weiteres Tabellenblatt anzufügen, muss der ExcelWriter im append-Mode `mode='a'` betrieben werden:

Zunächst erzeugen wir eine DataFrame durch anwenden der `describe()`-Methode: 

```python
df_describe = df8.describe()
print(type(df_describe))
df_describe
```

```python
Excel_xlsx_mehrere_Tabellen_Dateiname = r'df_mehrere_Tabellen_.xlsx'
with pd.ExcelWriter(Excel_xlsx_mehrere_Tabellen_Dateiname,mode='a') as writer:  
    df_describe.to_excel(writer, sheet_name='Sheet_name_3_describe')
```

Um die erste Spalte als Index auszuwählen, verwenden wir den Parameter `index_col=0`; die Spalte hat keinen Name bekommen und kann daher nur über den Integer Index angesprochen werden:

```python
df9 = pd.read_excel(Excel_xlsx_mehrere_Tabellen_Dateiname,sheet_name='Sheet_name_3_describe',index_col=0)
df9
```

# "Pickle" (Serialisierung) 

Die Serialisierung ([Wikipedia-Serialisierung](https://de.wikipedia.org/wiki/Serialisierung)) ist in der Informatik eine Abbildung von strukturierten Daten auf eine sequenzielle Darstellungsform. Serialisierung wird hauptsächlich für die Speicherung von Objekten in Dateien verwendet. 

- Serialisierung ("Pickling") - _Pandas DataFrame_ serialisieren und in eine Datei schreiben [`.to_pickle()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.to_pickle.html)
- Deserialisierung  ("unpickling") - _Pandas DataFrame_ aus der Datei wiederherstellen
[`pd.read_pickle()`](https://pandas.pydata.org/docs/reference/api/pandas.read_pickle.html)




```python
Pickle_Dateiname = r'df.pkl'
df.to_pickle(Pickle_Dateiname)
```

```python
df_pickle = pd.read_pickle(Pickle_Dateiname)
df_pickle
```

# JSON

JSON - JavaScript Object Notation - [Wikipedia](https://de.wikipedia.org/wiki/JSON)

#### JSON-Datei schreiben
[`.to_json()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.to_json.html)



DataFrame - Parameter `orient=` (orientation).
- default 'columns'
- verfügbare Werte: 'split', 'records', 'index', 'columns', 'values', 'table'


#### JSON-Datei lesen

[`pd.read_json()`](https://pandas.pydata.org/docs/reference/api/pandas.read_json.html)



#### Orientierung - default: `orient='columns'`

`{column -> {index -> value}}`

```python
JSON_columns_Dateiname = r'df_columns.json'
df.to_json(JSON_columns_Dateiname)
```

```python
df_json_columns = pd.read_json(JSON_columns_Dateiname)
df_json_columns
```

#### Orientierung - `orient='records'`

`[{column -> value}, ... , {column -> value}]`

der Index der Zeilen wird nicht mit abgespeichert.

```python
JSON_records_Dateiname = r'df_records.json'
df.to_json(JSON_records_Dateiname,orient='records')
```

```python
df_json_records = pd.read_json(JSON_records_Dateiname)
df_json_records
```

### JSON Datei in Python einlesen:

```python
import json
```

#### Orientierung - default: `orient='columns'`

`{column -> {index -> value}}`

```python
with open(JSON_columns_Dateiname) as f:
    json_data1 = json.load(f)
json_data1
```

```python
json_data1['Spalte1']['Zeile2']
```

#### Orientierung - `orient='records'`

`[{column -> value}, ... , {column -> value}]`

```python
with open(JSON_records_Dateiname) as f:
    json_data2 = json.load(f)
json_data2
```

```python
json_data2[1]['Spalte1']
```

```python
from sqlalchemy import create_engine 
```

#### Datenbank temporär im Speicher anlegen

Verbindung zur Datenbank erstellen mit:
`engine = create_engine('sqlite://')`

```python
engine = create_engine('sqlite:///')
```

Mit der [`.to_sql()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.to_sql.html)-Methode wird ein
_Pandas DataFrame_ als Tabelle in die Datenbank geschrieben.

Der erste Parameter ist der Name der Tabelle in der Datenbank.

Der zweite Parameter ist die Verbindung zur Datenbank.


```python
df.to_sql('meine_df_Tabelle',engine)
```

Mit der Funktion [`pd.read_sql()`](https://pandas.pydata.org/docs/reference/api/pandas.read_sql.html)
wird eine Tabelle aus der Datenbank in einen _Pandas DataFrame_ gelesen:

Erster Parameter ist der Tabellenname in der Datenbank.

Mit dem Paratemer `con=engine` wird die Verbindung zur Datenbank angegeben.

Mit dem Parameter `index_col='my_index'` wird die Spalte angegeben, die als Index des DataFrame ausgewähl werden soll.

```python
sqldf = pd.read_sql('meine_df_Tabelle',con=engine,index_col='my_index')
sqldf
```

#### SQL-Abfrage
Zugriff auf die Tabelle in der Datenbank mit SQL-Befehlen:

Zunächst wird mit der `.connect()`-Methode eine "Connection" zur Datenbank-Engine angelegt:

```python
connection = engine.connect()
```

```python
from sqlalchemy import text
```

Der SQL Query String kann der `.execute()` Methode nicht einfach übergeben werden. 
<br>
Durch die Funktion `text()` wird die Zeichenkette in eine ausführes Objekt umgewandelt.

```python
connection.execute(text("SELECT * FROM meine_df_Tabelle")).fetchall()
```

Eine weitere Abfrage:

```python
connection.execute(text("SELECT my_index,Spalte1 FROM meine_df_Tabelle")).fetchall()
```

Wenn wir mit den Abfragen fertig sind, schließen wir die Connection zur Datenbank.

```python
connection.close()
```

Dies geht aber auch aus Pandas heraus mit der
[`.read_sql_query()`](https://pandas.pydata.org/docs/reference/api/pandas.read_sql_query.html)-Funktion:

```python
sqldf = pd.read_sql_query("SELECT my_index,Spalte1 FROM meine_df_Tabelle",con=engine,index_col='my_index')
sqldf
```

oder der [`.read_sql()`](https://pandas.pydata.org/docs/reference/api/pandas.read_sql.html)- Methode:

```python
sqldf = pd.read_sql("SELECT my_index,Spalte1 FROM meine_df_Tabelle",con=engine,index_col='my_index')
sqldf
```

#### Datenbank in einer lokalen Datei anlegen

`sqlite://<nohostname>/<path>`
   
- relativer Pfad: `engine = create_engine('sqlite:///my_sqlite.db')`
- absoluter Pfad: `engine = create_engine('sqlite:///C:\\path\\to\\my_splite.db')`

```python
engine2 = create_engine('sqlite:///my_sqlite.db')
```

Durch den Parameter `if_exists=` wird das Verhalten festgelegt, falls die Tabelle schon existiert:
- _'fail'_ -> Fehler 'ValueError' ausgeben
- _'replace'_ -> die Werte der alten Tabelle werden verworfen und durch die neuen Werte ersetzt
- _'append'_ -> die neue Werte werden zu den existierenden Werten hinzugefügt.


```python
df.to_sql('meine_df_Tabelle',engine2, if_exists='replace')
```

```python
sqldf2 = pd.read_sql('meine_df_Tabelle',con=engine2,index_col='my_index')
sqldf2
```
