### Pandas - Zeitpunkt - `pd.Timestamp()`

[`pd.Timestamp()`](https://pandas.pydata.org/docs/reference/api/pandas.Timestamp.html)

spezifische Hilfe abrufen 

```python
#pd.Timestamp?
```

#### Zeitpunkt aus Zeichenkette nur mit Datum erstellen

```python
pd.Timestamp("2021-03-19")
```

Datentyp `pd.Timestamp`

```python
type(pd.Timestamp("2021-03-19"))
```

Type überprüfen

```python
isinstance(pd.Timestamp("2021-03-19"),pd.Timestamp)
```

#### Zeitpunkt aus Zeichenkette mit Datum und Uhrzeit erstellen

```python
pd.Timestamp("2021-03-19 18:13")
```

#### Angabe der einzelnen Bestandteile: Jahr, Monat, Tag, Stunde, Minute, Sekunde und Microsekunden

Unix-epoch: Sekunden nach 1.1.1970 
[Wikipedia-Unixzeit](https://de.wikipedia.org/wiki/Unixzeit)
[Wikipedia-Unix time](https://en.wikipedia.org/wiki/Unix_time) 

```python
pd.Timestamp(year=2017, month=1, day=1, hour=12, minute=5, second=10,  microsecond=500000)
```

#### Angabe als Sekunden nach 1.1.1970 00:00  `unit='s'`

```python
pd.Timestamp(0,unit='s')
```

```python
pd.Timestamp(1513393355.5, unit='s')
```

#### Angabe in Milli-Sekunden `unit='ms'`

```python
pd.Timestamp(1513393355.5*1000, unit='ms')
```

#### Angabe in Mikro-Sekunden `unit='us'`

```python
pd.Timestamp(1513393355.5*1000*1000, unit='us')
```

#### Angabe in Mikro-Sekunden `unit='ns'`

```python
pd.Timestamp(1513393355.5*1000*1000*1000, unit='ns')
```

### Pandas - Zeitbereich `pd.date_range()`

Die Funktion [`pd.date_range()`](https://pandas.pydata.org/docs/reference/api/pandas.date_range.html)
definiert einen Zeitbereich und gibt ein `DatetimeIndex`-Objekt (An immutable container for datetimes)
zurück.

spezifische Hilfe abrufen

```python
#pd.date_range?
```

#### Angabe eines Zeitbereichs durch ein Start- und ein Enddatum

```python
pd.date_range("2021-03-02","2021-03-06")
```

#### Angabe eines Zeitbereichs durch ein Startdatum und die Anzahl von Tagen

```python
pd.date_range("2021-03-02", periods=5)
```

#### Angabe eines Zeitbereichs durch ein Enddatum und die Anzahl von Tagen

```python
pd.date_range(end="2021-03-06", periods=5)
```

weitere Datenobjekte neben `DatetimeIndex` - An immutable container for datetimes
- `timedelta_range` : Return a fixed frequency TimedeltaIndex.
- `period_range` : Return a fixed frequency PeriodIndex.
- `interval_range` : Return a fixed frequency IntervalIndex.
